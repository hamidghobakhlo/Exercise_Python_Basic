# Copy

2. Deep Copy

```python
t1 = (4,8,12,16)
t2 = tuple(t1)
print(t1, t2)
```

### list

```python
list1 = [2,5,6,7,8,]
list2 = [2,5,6,7,8,]
list3 = list1

print(list1 == list2)
print(list1 is list2)
print(list1 is list3)

print(id(list1))
print(id(list3))
print(id(list2))

list1.append(3000)
print(list1, list3)
```

```python
list1 = [2,5,6,7,8,]
list2 = [2,5,6,7,8,]
list3 = list1[:] # Shallow copy

list1.append(3000)
print(list1, list3)
```

```python
list1 = [2,5,6,7,8, [2, 4, 8]] 
print(list1[5][0])
list3 = list1[:] # Shallow copy
list1[5][0] = 200
print(list1, list3)
```

## DeepCopy

```python
import copy

list1 = [2,5,6,7,8, [2, 4, 8]] 
print(list1[5][0])
list3 = copy.deepcopy(list1)
list1[5][0] = 200
print(list1, list3)
```

### Dictionary

```python
dict1 = {
    "a" : 2,
    "b" : {"c" : 20, "d" : 5}
}
```

```python
dict2 = dict(dict1) # Copy ===> dict(), set(), list(), tuple() # Shallow copy
print(dict2)
dict2["b"]["C"] = 200
print(dict2)
print(dict1)
```

```python
import copy 

dict1 = {
    "a" : 2,
    "b" : {"c" : 20, "d" : 5}
}

dict2 = copy.deepcopy(dict1)
print(dict2)
dict2["b"]["C"] = 200
print(dict2)
print(dict1)
```

### Set

```python
set1 = {4, 5, 6, 8}
set2 = set(set1)
set2.add(30)
print(set1, set2)
```
