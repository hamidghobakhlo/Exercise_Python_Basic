```python
# Recursive Function - Factorial
# f(4)
# f(4) = 4 * f(3)
# f(3) = 3 * f(2)
# f(2) = 2 * f(1)
# f(1) = 1 
# f(n) ===> n * f(n-1)
```

```python
# None Recursive Function - Factorial
def factorial1(num):
    f = 1
    for i in range(1, num+1):
        f *= i
    print(f)
```

```python
factorial1(5)
```

```python
# Recursive Function - Factorial
def factorial2(num):
    if num == 1 or num == 0:
        return 1
    else:
        return num * factorial2(num - 1)
```

```python
factorial2(5)
```

### Binary Search

```python
def binary_search(list1, x, low, high):
    if high >= low:
        mid = (low + high) //2
        if list1[mid] == x:
            return mid
        elif list1[mid] > x:
            return binary_search(list1, x, low, mid - 1)
        else:
            return binary_search(list1, x, mid + 1, high)
    else:
        return -1
```

```python
list1 = [2,5,6,10,14]
x = 50
res = binary_search(list1, x, 0, len(list1)-1)
print(res)
```
